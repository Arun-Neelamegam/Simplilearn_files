##### **Steps to be followed:**

1. **Store customer information:**
    - Create a list of customer names
    - Store each customer's details (name, age, and location) as tuples inside a list
    - Use a dictionary to store feedback from each customer
    - Print the stored customer details

2. **Analyze feedback:**
    - Create separate sets of positive and negative keywords
    - Create a dictionary to store the classification results
    - Loop through each customer's feedback
    - Convert the feedback to lowercase and remove punctuation for keyword matching
    - Classify the feedback:
        - Mixed if both positive and negative keywords are found
        - Positive if only positive keywords are found
        - Negative if only negative keywords are found
        - Neutral if neither is found
    - Print the classified feedback

3. **Generate a summary report:**
    - Use a list comprehension to list customers classified as Positive or Mixed
    - Print the positive customer list
    - Store unique lowercase words from all feedback messages using a set, excluding punctuation
    - Display the unique words in alphabetical order

4. **Organize and display data:**
    - Create positive and negative customer sets, including Mixed customers in both
    - Find customers with positive or negative feedback (union)
    - Find customers with both positive and negative feedback (intersection)

    - Display the results

**Step 1:** Store customer information

```python
# Create a list of customer names
customer_names = ["Alice", "Bob", "Charlie", "David"]

# Store each customer's details (name, age, and location) as tuples inside a list.
customer_details = [("Alice", 28, "New York"),
                    ("Bob", 34, "Los Angeles"),
                    ("Charlie", 25, "Chicago"),
                    ("David", 40, "Miami")]

# Use a dictionary to store feedback from each customer.
customer_feedback = {
    "Alice": "The service was excellent and very professional.",
    "Bob": "I had a bad experience with customer support.",
    "Charlie": "Good quality but delivery was slow.",
    "David": "The product was poor and not worth the price."
}

# Print stored customer details
print("Customer Details:")
for details in customer_details:
    print(details)
```

**Step 2:** Analyze feedback

```python
import string

# Keywords for classification
positive_words = {"good", "excellent", "great", "amazing", "professional"}
negative_words = {"bad", "poor", "terrible", "slow", "worst"}

# Dictionary to store classification results
feedback_classification = {}

# Loop through customer feedback and classify
for customer, feedback in customer_feedback.items():
    cleaned_feedback = feedback.lower()
    for punctuation in string.punctuation:
        cleaned_feedback = cleaned_feedback.replace(punctuation, " ")
    words = set(cleaned_feedback.split())

    if (words & positive_words) and (words & negative_words):
        feedback_classification[customer] = "Mixed"
    elif words & positive_words:
        feedback_classification[customer] = "Positive"
    elif words & negative_words:
        feedback_classification[customer] = "Negative"
    else:
        feedback_classification[customer] = "Neutral"

# Print classified feedback
print("\nCustomer Feedback Classification:")
for customer, category in feedback_classification.items():
    print(f"{customer}: {category}")
```

**Step 3:** Generate a summary report

```python
import string

# List customers with positive feedback, including mixed feedback
positive_customers = [
    customer
    for customer, category in feedback_classification.items()
    if category in ("Positive", "Mixed")
]

print("\nCustomers with positive feedback:")
print(positive_customers)

# Collect unique words without punctuation
unique_words = set()

for feedback in customer_feedback.values():
    cleaned_feedback = feedback.lower()
    for punctuation in string.punctuation:
        cleaned_feedback = cleaned_feedback.replace(punctuation, " ")
    unique_words.update(cleaned_feedback.split())

print("\nUnique words in customer feedback:")
print(sorted(unique_words))
```

**Step 4:** Organize and display data

```python
# Include mixed feedback in both sets
positive_customers_set = {
    customer
    for customer, category in feedback_classification.items()
    if category in ("Positive", "Mixed")
}

negative_customers_set = {
    customer
    for customer, category in feedback_classification.items()
    if category in ("Negative", "Mixed")
}

# Perform set operations
print("\nCustomers with positive or negative feedback:")
print(sorted(positive_customers_set | negative_customers_set))

print("\nCustomers with both positive and negative feedback:")
print(sorted(positive_customers_set & negative_customers_set))

print("\nCustomers with only negative feedback:")
print(sorted(negative_customers_set - positive_customers_set))
```
